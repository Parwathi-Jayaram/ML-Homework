```python
import numpy as np
```

```python
def dotproduct(u,v):
    assert u.shape[0]==v.shape[0]
    n=u.shape[0]
    result=0
    for i in range(n):
        result=result+u[i]*v[i]
    return result
```

```python
v=np.array([
    [1,1,2],
    [0,0.5,1],
    [0,2,1],
    [2,1,0],
])
```

```python
u=np.array([
    [2,4,5,6],
    [1,2,1,2],
    [3,1,2,1],
    ])
```

```python
def matrix_vector(u,v):
    assert u.shape[1]==v.shape[0]
    k=u.shape[0]
    result=np.zeros(k)
    for i in range(k):
        result[i]=u
    return result
```

```python
def matrix_matrix(u,v):
    assert u.shape[1]==v.shape[0]
    k=u.shape[0]
    result=np.zeros((u.shape[0],v.shape[1]))
    for i in range(k):
        ui=u[:,i]
        result[:,i]=matrix_vector(u,v[:,i])
    return result
```

```python
matrix_matrix(u,v)
```
